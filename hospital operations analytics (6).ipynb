<h3 style="color:#236B45; font-size:22px;">🔍 MISSING VALUES CHECK</h3>

</div>

```python
df.isnull().sum()
```

```python
# Fill missing values in categorical columns with mode
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])

# Fill missing values in numerical columns with median
df['Length_of_Stay'] = df['Length_of_Stay'].fillna(df['Length_of_Stay'].median())
df['Oxygen_Units_Used'] = df['Oxygen_Units_Used'].fillna(df['Oxygen_Units_Used'].median())
```

<h3 style="color:#236B45; font-size:22px;">🔎 OUTLIER DETECTION AND HANDLING</h3>

</div>

```python
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = ((df[col] < Q1 - 1.5 * IQR) | (df[col] > Q3 + 1.5 * IQR)).sum()
    print(col, ":", outliers)
```

```python
for col in ['Length_of_Stay', 'Oxygen_Units_Used']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    df[col] = df[col].clip(lower=lower_limit, upper=upper_limit)
```

```python
for col in ['Length_of_Stay', 'Oxygen_Units_Used']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    outliers = ((df[col] < Q1 - 1.5 * IQR) | 
                (df[col] > Q3 + 1.5 * IQR)).sum()
    
    print(col, ":", outliers)
```
